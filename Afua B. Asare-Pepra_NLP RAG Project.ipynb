## Question Answering using LLM with Prompt Engineering

```python
# this code defines a function to generate responses from the LLM
def generate_llama_response(query, instruction="Answer the following question clearly and concisely."):

    system_message = f"[INST]<<SYS>>\n{instruction}\n<</SYS>>[/INST]"

    prompt = f"{system_message} {query}"

    response = llm(
        prompt=prompt,
        max_tokens=1024,
        temperature=0.7,
        top_p=0.95,
        repeat_penalty=1.1,
        top_k=50,
        stop=['</s>'],
        echo=False,
        seed=42,
    )

    return response["choices"][0]["text"].strip()
```

### System and User Prompt Template

```python
qna_system_message = """You are an AI medical assistant.
Your task is to provide clear, concise, and accurate medical information based *only* on the provided context.
Do not invent information or provide medical advice.
If the answer is not in the context, state that you cannot provide it based on the given information.
"""

qna_user_message_template = """Based on the following context, answer the question:

Context: {context}

Question: {question}"""
```

### Response Function

```python
def generate_rag_response(user_input, k=3, max_tokens=1024, temperature=0, top_p=0.95, top_k=50):
    global qna_system_message, qna_user_message_template, vectorstore

    # Create a temporary retriever with the specified k
    dynamic_retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": k}
    )

    # Retrieve relevant document chunks using the dynamic retriever
    relevant_document_chunks = dynamic_retriever.invoke(input=user_input)
    context_list = [d.page_content for d in relevant_document_chunks]

    # Combine document chunks into a single context
    context_for_query = ". ".join(context_list)

    user_message = qna_user_message_template.replace('{context}', context_for_query)
    user_message = user_message.replace('{question}', user_input)

    prompt = qna_system_message + '\n' + user_message

    # Generate the response
    try:
        response = llm(
                  prompt=prompt,
                  max_tokens=max_tokens,
                  temperature=temperature,
                  top_p=top_p,
                  top_k=top_k
                  )

        # Extract and print the model's response
        response = response['choices'][0]['text'].strip()
    except Exception as e:
        response = f'Sorry, I encountered the following error: \n {e}'

    return response
```

## Output Evaluation

Let us now use the LLM-as-a-judge method to check the quality of the RAG system on two parameters - retrieval and generation. We illustrate this evaluation based on the answeres generated to the question from the previous section.

- We are using the same Mistral model for evaluation, so basically here the llm is rating itself on how well he has performed in the task.

```python
# this code evaluates the groundedness of the medical response
groundedness_rater_system_message = "You are assessing whether a medical response is grounded. Determine whether the facts made are supported by the context provided. Rate on a scale of 1-5 (1: off tangent, 5: accurate and supported by context) "
```

```python
# this code determines the relevance of the medical response
relevance_rater_system_message = "You are determining how relevant a medical response is. Rate on a scale of 1-5 (1: irrelevant, 5: highly relevant) "
```

```python
user_message_template = """Context: {context}

Question: {question}

Answer: {answer}"""
```

```python
def generate_ground_relevance_response(user_input,k=3,max_tokens=128,temperature=0,top_p=0.95,top_k=50):
    global qna_system_message,qna_user_message_template, vectorstore

    # Create a temporary retriever with the specified k
    dynamic_retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": k}
    )
    # Retrieve relevant document chunks
    relevant_document_chunks = dynamic_retriever.invoke(input=user_input)
    context_list = [d.page_content for d in relevant_document_chunks]
    context_for_query = ". ".join(context_list)

    # Combine user_prompt and system_message to create the prompt
    prompt = f"""[INST]{qna_system_message}\n
                {'user'}: {qna_user_message_template.format(context=context_for_query, question=user_input)}
                [/INST]"""

    response = llm(
            prompt=prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    answer =  response["choices"][0]["text"]

    # Combine user_prompt and system_message to create the prompt
    groundedness_prompt = f"""[INST]{groundedness_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    # Combine user_prompt and system_message to create the prompt
    relevance_prompt = f"""[INST]{relevance_rater_system_message}\n
                {'user'}: {user_message_template.format(context=context_for_query, question=user_input, answer=answer)}
                [/INST]"""

    response_1 = llm(
            prompt=groundedness_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    response_2 = llm(
            prompt=relevance_prompt,
            max_tokens=max_tokens,
            temperature=temperature,
            top_p=top_p,
            top_k=top_k,
            stop=['INST'],
            )

    return response_1['choices'][0]['text'],response_2['choices'][0]['text']
```
